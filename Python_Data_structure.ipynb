DATA TYPES AND STRUCTURE QUESTIONS


1.  What are data structures, and why are they important?
- Data structures are special ways of organizing, storing, and managing data so that it can be used efficiently by programs.
They determine how data is arranged in memory and what operations can be performed quickly

2.  Explain the difference between mutable and immutable data types with examples.
- Mutable data types can be changed after they are created.
You can modify, add, or remove elements without creating a new object in memory.
ex: list,dict,set.
Immutable data types cannot be changed after they are created.
If you try to modify them, Python creates a new object in memory.
ex: int,float,str,tuple,bool.

3.  What are the main differences between lists and tuples in Python?
- Lists are mutable, meaning their elements can be changed, added, or removed after creation Because of this flexibility, lists are commonly used for data that needs to be modified during the program, such as dynamic datasets or user inputs.
Tuples are immutable, which means once a tuple is created, none of its elements can be changed. This immutability makes tuples faster, more memory-efficient, and safer for storing fixed data, such as configuration values or coordinates that should not be altered.

4. Describe how dictionaries store data
- Dictionaries in Python store data using a key value structure.

5. Why might you use a set instead of a list in Python.
- A set is used instead of a list in Python when you need unique items, fast membership checking, and unordered data handling.

6.  What is a string in Python, and how is it different from a list?
- A string in Python is an immutable sequence of characters.Strings are primarily used to represent text such as words, sentences, or symbols.A list, on the other hand, is a mutable sequence that can store elements of any data type, including numbers, strings, or even other lists.

7.  How do tuples ensure data integrity in Python
- Tuples ensure data integrity by being immutable.

8.  What is a hash table, and how does it relate to dictionaries in Python?
- A hash table is a data structure that stores key value pairs and uses a hash function to quickly find the memory location where each value should be stored.

In Python, dictionaries are implemented internally using hash tables, which is why dictionary operations like lookup, insert, and delete are very fast.

9.  Can lists contain different data types in Python?
- Yes, lists can contain different data types in Python.

10. Explain why strings are immutable in Python.
- Strings are immutable in Python because they are stored as fixed sequences of characters in memory, and Python does not allow in-place modification of those memory blocks.

11.  What advantages do dictionaries offer over lists for certain tasks
- Dictionaries are advantageous when you need fast, direct access to data using meaningful keys instead of numeric positions.
They store data as key value pairs, so you can quickly retrieve, update, or check items by their key.

12.  Describe a scenario where using a tuple would be preferable over a list
- A tuple would be preferable over a list when you want to store a fixed collection of values that should not change, such as the coordinates of a point (x, y) in a 2D space or the RGB values of a color (255, 0, 0).

13. How do sets handle duplicate values in Python?
- In Python, sets automatically remove duplicate values. When you add elements to a set, any repeated values are ignored, so each element in a set is unique.

14.  How does the “in” keyword work differently for lists and dictionaries?
- In list it checks if a value exists anywhere in the list.
 In dictionaries it checks if a key exists in the dictionary, not the values.

15. Can you modify the elements of a tuple? Explain why or why not.
- No, you cannot modify the elements of a tuple in Python because tuples are immutable.

Immutability means once a tuple is created, its contents cannot be changed, added, or removed.

This ensures data integrity and allows tuples to be used as dictionary keys or stored in sets, which require elements to be hashable.

16.  What is a nested dictionary, and give an example of its use case?
- A nested dictionary in Python is a dictionary that contains another dictionary as a value for one or more keys. It allows you to represent hierarchical or structured data.
Use Case:
Nested dictionaries are useful for storing complex data, such as details of multiple students in a class, where each student has multiple attributes (age, grade, etc.).

17.  Describe the time complexity of accessing elements in a dictionary?
- In Python, accessing elements in a dictionary (using a key) has an average time complexity of O(1).

18. In what situations are lists preferred over dictionaries?
- Lists are preferred over dictionaries in situations where:Order matters, Indexed access, Homogeneous data.

19. Why are dictionaries considered unordered, and how does that affect data retrieval?
- Dictionaries in Python are considered unordered  because the items are stored based on their hash values, not in the order they were added.
Effect on data retrieval:

You cannot rely on the order of keys or values when iterating through a dictionary.

Retrieval by key is still fast (O(1) average) because dictionaries use a hash table internally.

20.  Explain the difference between a list and a dictionary in terms of data retrieval.
- 1. List

Elements are accessed by index (position).

Retrieval requires sequential access if you search by value:

Average time complexity: O(n)
  
  2. Dictionary

Elements are accessed by key.

Retrieval is direct via the key:

Average time complexity: O(1)
